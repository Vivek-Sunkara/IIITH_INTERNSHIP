```python
import subprocess

# Input video file
input_video = "input.mp4"

# Output image pattern
output_images = "frame_%04d.png"

# FFmpeg command
command = [
    "ffmpeg",
    "-i", input_video,
    "-r", "1",              # 1 frame per second
    output_images
]

# Run command
subprocess.run(command)

print("Frames extracted successfully!")
```
