```python
import os
import subprocess

# ==============================
# 1. CONFIGURATION
# ==============================
input_video = "input.mp4"   # <-- your video file
clip_video = "clip.mp4"
frames_folder = "frames"
output_video = "output.mp4"

start_time = "00:00:10"   # change if needed
duration = "60"           # 1 minute

# ==============================
# 2. CUT 1-MINUTE CLIP
# ==============================
print("Cutting 1-minute clip...")

subprocess.run([
    "ffmpeg",
    "-ss", start_time,
    "-i", input_video,
    "-t", duration,
    "-c", "copy",
    clip_video,
    "-y"
])

# ==============================
# 3. EXTRACT FRAMES (30 FPS)
# ==============================
print("Extracting frames...")

os.makedirs(frames_folder, exist_ok=True)

subprocess.run([
    "ffmpeg",
    "-i", clip_video,
    "-vf", "fps=30",
    f"{frames_folder}/frame_%04d.png"
])

# ==============================
# 4. REBUILD VIDEO FROM FRAMES
# ==============================
print("Rebuilding video from frames...")

subprocess.run([
    "ffmpeg",
    "-framerate", "30",
    "-i", f"{frames_folder}/frame_%04d.png",
    "-c:v", "libx264",
    "-pix_fmt", "yuv420p",
    output_video,
    "-y"
])

print("✅ Done!")
print(f"Final video saved as: {output_video}")
```
