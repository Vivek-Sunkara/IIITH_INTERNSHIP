```python
from ultralytics import YOLO
import cv2
import os
```

```python
def detect_video_yolo(input_video_path, output_video_path='output_video.mp4', model_path='yolov8m.pt'):
    """
    Detect objects in video using YOLO model and save annotated video
    
    Args:
        input_video_path: Path to input video file
        output_video_path: Path to save output video (default: 'output_video.mp4')
        model_path: Path to YOLO model (default: 'yolov8m.pt')
    """
    
    # Load YOLO model
    print(f"Loading model: {model_path}")
    model = YOLO(model_path)
    
    # Open video file
    cap = cv2.VideoCapture(input_video_path)
    
    if not cap.isOpened():
        print(f"Error: Cannot open video file {input_video_path}")
        return
    
    # Get video properties
    fps = int(cap.get(cv2.CAP_PROP_FPS))
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    
    print(f"Video properties - FPS: {fps}, Resolution: {width}x{height}, Total frames: {total_frames}")
    
    # Define video writer
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(output_video_path, fourcc, fps, (width, height))
    
    if not out.isOpened():
        print(f"Error: Cannot create output video {output_video_path}")
        cap.release()
        return
    
    frame_count = 0
    
    # Process video frames
    while True:
        ret, frame = cap.read()
        
        if not ret:
            break
        
        frame_count += 1
        
        # Run YOLO inference
        results = model.predict(source=frame, conf=0.5, verbose=False)
        
        # Draw annotations on frame
        annotated_frame = results[0].plot()
        
        # Write frame to output video
        out.write(annotated_frame)
        
        # Print progress
        if frame_count % 30 == 0:
            print(f"Processed {frame_count}/{total_frames} frames")
    
    # Release resources
    cap.release()
    out.release()
    
    print(f"\nDetection completed!")
    print(f"Output video saved to: {output_video_path}")
    print(f"Total frames processed: {frame_count}")
```

```python
# Specify your input video path here
input_video_path = '00096.mp4'  # Change this to your video file path

# Specify output video path
output_video_path = 'output_video.mp4'
```

```python
# Check if input video exists
if os.path.exists(input_video_path):
    detect_video_yolo(input_video_path, output_video_path, model_path='yolov8m.pt')
else:
    print(f"Error: Input video file '{input_video_path}' not found!")
    print("Please update the 'input_video_path' variable with the correct path to your video.")
```

```python
# Check if output video was created
if os.path.exists(output_video_path):
    file_size_mb = os.path.getsize(output_video_path) / (1024 * 1024)
    print(f"Output video size: {file_size_mb:.2f} MB")
else:
    print("Output video was not created. Please check the error messages above.")
```
